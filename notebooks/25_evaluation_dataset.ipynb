# 25 — Evaluation Dataset Setup

Loads and validates the IoT security eval set used for research experiments.

**Dataset:** `datasets/evaluation/iot_security_eval_v1.json` (50 questions)

Use this notebook to inspect categories, facet demand, and sample questions before running baselines in notebook 26.

```python
import os
import sys

project_root = os.path.abspath("..")

if project_root not in sys.path:
    sys.path.insert(0, project_root)
```

```python
from rag_icot.evaluation import load_eval_dataset, summarize_dataset

dataset_path = os.path.join(
    project_root,
    "datasets",
    "evaluation",
    "iot_security_eval_v1.json",
)

questions = load_eval_dataset(dataset_path)
summary = summarize_dataset(questions)

print("Loaded", summary["count"], "questions")
print("By category:", summary["by_category"])
print("By difficulty:", summary["by_difficulty"])
print("Facet demand:", summary["facet_demand"])
```

```python
# Preview a few questions from each major category
from collections import defaultdict

by_cat = defaultdict(list)
for q in questions:
    by_cat[q.category].append(q)

for category, items in by_cat.items():
    print("\n===", category, f"({len(items)}) ===")
    for q in items[:2]:
        print(f"{q.id} [{q.difficulty}] facets={q.required_facets}")
        print(" ", q.question)
```

```python
# Faithfulness / refusal questions (important for hallucination eval)
faithfulness = [q for q in questions if q.category == "faithfulness"]

for q in faithfulness:
    print(f"{q.id}: {q.question}")
    print("  gold_notes:", q.gold_notes)
    print()
```

```python
# Smoke-check: every question object validates and has unique IDs
ids = [q.id for q in questions]
assert len(ids) == len(set(ids)), "Duplicate question IDs"
assert len(questions) == 50, "Expected 50 questions"
print("Validation OK")
```
