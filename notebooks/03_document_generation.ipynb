```python
import sys
sys.path.append("..")


from rag_icot.components.data_ingestion import DataIngestion
```

```python
ingestion = DataIngestion()
```

```python
df = ingestion.load_iot23(
    "../datasets/iot23/CTU-IoT-Malware-Capture-1-1/bro/conn.log.labeled"
)
```

```python
from rag_icot.components.document_generator import DocumentGenerator

generator = DocumentGenerator()
```

```python
documents = []

for _, row in df.iterrows():
    doc = generator.generate_document(row)
    documents.append(doc)

print(len(documents))
```

```python
import json
import os

os.makedirs("artifacts", exist_ok=True)

with open("../artifacts/documents.json", "w", encoding="utf-8") as f:
    json.dump(documents, f, indent=4)

print("Documents saved successfully!")
```
