```python
from rag_icot.components.retriever import Retriever
from rag_icot.components.context_manager import ContextManager

retriever = Retriever()

context = ContextManager()

generator = tg.ThoughtGenerator()
```

```python
query = "How to protect my network?"
```

```python
results = retriever.retrieve(
    query,
    k=5
)
```

```python
context.add_documents(

    results["ids"][0],

    results["documents"][0],

    results["metadatas"][0]

)
```

```python
context_text = ""

for doc in context.get_documents():

    context_text += doc["text"]

    context_text += "\n\n"
```

```python
thought = generator.generate(

    query,

    context_text

)
```
