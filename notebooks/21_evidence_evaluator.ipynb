```python
import os
import sys

project_root = os.path.abspath("..")

if project_root not in sys.path:
    sys.path.insert(0, project_root)
```

```python
from rag_icot.components.retriever import Retriever
from rag_icot.components.context_manager import ContextManager
from rag_icot.components.evidence_evaluator import EvidenceEvaluator
```

```python
retriever = Retriever()

context = ContextManager()

evaluator = EvidenceEvaluator()
```

```python
question = "How can I protect my network from Mirai botnet attacks?"

results = retriever.retrieve(
    question,
    k=5
)

context.add_documents(
    results["ids"][0],
    results["documents"][0],
    results["metadatas"][0]
)
```

```python
context_text = ""

for doc in context.get_documents():

    context_text += doc["text"]

    context_text += "\n\n"
```

```python
evaluation = evaluator.evaluate(
    question,
    context_text
)

print(evaluation)
```
