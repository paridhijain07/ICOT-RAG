```python
import os
import sys

project_root = os.path.abspath("..")

if project_root not in sys.path:
    sys.path.insert(0, project_root)
```

```python
from rag_icot.components.reasoning_step import ReasoningStep
from rag_icot.components.retriever import Retriever
from rag_icot.components.context_manager import ContextManager
```

```python
retriever = Retriever()

context = ContextManager()

reasoning = ReasoningStep()
```

```python
question = "How can I protect my IoT devices from Mirai botnet attacks?"
```

```python
results = retriever.retrieve(
    question,
    k=5
)
```

```python
context.add_documents(
    results["ids"][0],
    results["documents"][0],
    results["metadatas"][0]
)
```

```python
context_text = ""

for doc in context.get_documents():

    context_text += doc["text"]

    context_text += "\n\n"
```

```python
result = reasoning.run(
    question,
    context_text
)
```
