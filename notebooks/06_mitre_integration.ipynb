```python
selected_types = {
    "attack-pattern",
    "malware",
    "tool",
    "intrusion-set",
    "course-of-action"
}

selected_objects = []

for obj in mitre["objects"]:
    if obj.get("type") in selected_types:
        selected_objects.append(obj)

print(len(selected_objects))
```

```python
print(selected_objects[1])
```

```python
import re

def clean_text(text):
    if not text:
        return ""

    # Remove citations like (Citation: Microsoft...)
    text = re.sub(r"\(Citation:[^)]+\)", "", text)

    # Replace multiple newlines with a single newline
    text = re.sub(r"\n+", "\n", text)

    return text.strip()
```

```python
# documents = []

# for idx, obj in enumerate(selected_objects):

#     technique_id = ""

#     for ref in obj.get("external_references", []):
#         if ref.get("source_name") == "mitre-attack":
#             technique_id = ref.get("external_id", "")
#             break

#     kill_chain = ", ".join(
#         [
#             phase["phase_name"]
#             for phase in obj.get("kill_chain_phases", [])
#         ]
#     )

#     platforms = ", ".join(
#         obj.get("x_mitre_platforms", [])
#     )

#     data_sources = ", ".join(
#         obj.get("x_mitre_data_sources", [])
#     )

#     doc = {

#         "id": f"mitre_{idx}",

#         "source": "MITRE",

#         "document_type": obj.get("type"),

#         "text": f"""
# Technique ID: {technique_id}

# Name: {obj.get("name","")}

# Description:
# {obj.get("description","")}

# Platforms:
# {platforms}

# Kill Chain:
# {kill_chain}

# Detection:
# {obj.get("x_mitre_detection","")}

# Data Sources:
# {data_sources}
# """.strip(),

#         "metadata": {

#             "stix_id": obj.get("id"),

#             "created": obj.get("created"),

#             "modified": obj.get("modified")

#         }

#     }

#     documents.append(doc)

documents = []

for idx, obj in enumerate(selected_objects):

    technique_id = ""

    for ref in obj.get("external_references", []):
        if ref.get("source_name") == "mitre-attack":
            technique_id = ref.get("external_id", "")
            break

    doc = {

        "id": f"mitre_{idx}",

        "source": "MITRE",

        "document_type": obj.get("type"),

        "technique_id": technique_id,

        "name": obj.get("name", ""),

        "description": clean_text(
            obj.get("description", "")
        ),

        "platforms": obj.get(
            "x_mitre_platforms", []
        ),

        "kill_chain": [
            phase["phase_name"]
            for phase in obj.get(
                "kill_chain_phases", []
            )
        ],

        "detection": clean_text(
            obj.get(
                "x_mitre_detection", ""
            )
        ),

        "data_sources": obj.get(
            "x_mitre_data_sources", []
        ),

        "metadata": {

            "stix_id": obj.get("id"),

            "created": obj.get("created"),

            "modified": obj.get("modified")

        }

    }

    documents.append(doc)
```

```python
print(documents[0])
```

```python
import json

with open(
    "../artifacts/mitre_documents.json",
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        documents,
        f,
        indent=4,
        ensure_ascii=False
    )
```
