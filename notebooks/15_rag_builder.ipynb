```python
import os
import sys

project_root = os.path.abspath("..")

if project_root not in sys.path:
    sys.path.insert(0, project_root)
```

```python
from rag_icot.components.retriever import Retriever
from rag_icot.components.prompt_builder import PromptBuilder
from rag_icot.components.llm import GeminiLLM
```

```python
retriever = Retriever()

prompt_builder = PromptBuilder()

llm = GeminiLLM()
```

```python
query = "What are DDoS attacks?"
```

```python
results = retriever.retrieve(
    query,
    k=5
)
```

```python
documents = results["documents"][0]
```

```python
prompt = prompt_builder.build(
    query,
    documents
)
```

```python
print(prompt)
```

```python
answer = llm.generate(
    prompt
)
```
