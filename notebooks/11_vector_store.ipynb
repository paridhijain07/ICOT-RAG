```python
import os
import sys

project_root = os.path.abspath("..")

if project_root not in sys.path:
    sys.path.append(project_root)

print(project_root)
```

```python
from rag_icot.components.knowledge_base_builder import KnowledgeBaseBuilder
from rag_icot.components.embedding_builder import EmbeddingBuilder
from rag_icot.components.vector_store import VectorStore
```

```python
kb = KnowledgeBaseBuilder()

master_documents = kb.load_documents(
    "../artifacts/master_documents.json"
)

print(len(master_documents))
```

```python
builder = EmbeddingBuilder()

embeddings = builder.embed_documents(
    master_documents
)

print(type(embeddings))
print(embeddings.shape)
```

```python
builder.save_embeddings(
    embeddings,
    "../artifacts/master_embeddings.npy"
)
```

```python
loaded = builder.load_embeddings(
    "../artifacts/master_embeddings.npy"
)

print(type(loaded))
print(loaded.shape)
```

```python
store = VectorStore()

collection = store.create_collection()
```

```python
store.add_documents(
    master_documents,
    loaded
)
```

```python
print(collection.count())
```

```python
query = builder.embed(
    "Mirai botnet DDoS attack"
)

results = store.search(
    query,
    k=5
)
```
