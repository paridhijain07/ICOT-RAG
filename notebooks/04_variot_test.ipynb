```python
vulns = loader.get_vulnerabilities(5)
```

```python
records = []
```

```python
for v in vulns:

    clean_record = {
        "variot_id": v.get("id"),
        "cve": v.get("cve"),
        "title": v.get("title", {}).get("data"),
        "description": v.get("description", {}).get("data"),
        "threat_type": v.get("threat_type", {}).get("data")
    }

    records.append(clean_record)
```

```python
print(records[0])
```

```python
import json

with open(
    "../artifacts/variot_vulnerabilities.json",
    "w"
) as f:
    json.dump(records, f, indent=4)
```
