# 29 — LLM-as-Judge + Iteration Ablation

1. **Judge** vanilla vs `facet_icot` on the 10-question smoke set  
   Dimensions: reliability, relevance, technicality, friendliness (1–5)
2. **Ablation** on multi-facet questions: vanilla vs `max_iter=1` vs `max_iter=3`

Prefer running the script (handles pauses / saves artifacts):

```bash
python scripts/run_judge_and_iter_ablation.py
```

Outputs:
- `artifacts/evaluation/llm_judge_smoke.json`
- `artifacts/evaluation/llm_iter_ablation.json`

````python
import os
import sys
import json

project_root = os.path.abspath("..")
if project_root not in sys.path:
    sys.path.insert(0, project_root)
````

````python
judge_path = os.path.join(
    project_root, "artifacts", "evaluation", "llm_judge_smoke.json"
)
ablate_path = os.path.join(
    project_root, "artifacts", "evaluation", "llm_iter_ablation.json"
)

if os.path.exists(judge_path):
    judge = json.load(open(judge_path, encoding="utf-8"))
    print("Judge summary")
    print(json.dumps(judge.get("summary", {}), indent=2))
else:
    print("Missing", judge_path, "— run scripts/run_judge_and_iter_ablation.py")

if os.path.exists(ablate_path):
    ablate = json.load(open(ablate_path, encoding="utf-8"))
    print("\nIter ablation summary")
    print(json.dumps(ablate.get("summary", {}), indent=2))
else:
    print("Missing", ablate_path)
````
